```python
import xml.etree.ElementTree as et

tree = et.ElementTree(file='/menu.xml')
root = tree.getroot()

print(root.tag)
```

```python
type(root)
```

```python
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)
```

```python
print(len(root))      # menu의 하위 태그 수
print(len(root[0]))   # breakfast의 item 수
print(type(root[0]))
```

```python
print(root[0])
print(root[0].get("hours"))
print(root[0].keys())
print(root[0].items())
```

```python
lunch = root.find('lunch')

print(lunch)
print(lunch.get('hours'))
print(lunch.keys())
print(lunch.items())
print(type(lunch))
```

```python
print(root[0].findall('item'))

for item in root[0].findall('item'):
    print(item.get('price'))
    print(item.text)
```

```python
import json

j1 = {"name": "홍길동", "birth": "0525", "age": 30}

print(j1)
print(json.dumps(j1))
```

```python
print(json.dumps(j1, indent=2))

print(json.dumps([1, 2, 3]))
print(json.dumps((4, 5, 6)))
```

```python
d1 = json.dumps(j1)

print(d1)
print(json.loads(d1))
print(type(json.loads(d1)))
```

```python
myinfo = {
    "name": "홍길동",
    "birth": "0525",
    "age": 30
}

with open('./myinfo.json', 'w', encoding='utf-8') as f:
    json.dump(myinfo, f, ensure_ascii=False, indent=2)

with open('./myinfo.json', encoding='utf-8') as f:
    data = json.load(f)

print(type(data))
print(data)
```
